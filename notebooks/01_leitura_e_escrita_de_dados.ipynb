### Lendo e Salvando Arquivos CSV

Esta seção foca em como ler dados de arquivos CSV (Comma Separated Values) para DataFrames Pandas e como salvar DataFrames de volta em arquivos CSV. CSV é um formato comum para armazenar dados tabulares.

### 📖 Como funciona a leitura de CSV com Pandas?

O **Pandas** é a principal biblioteca do Python para análise de dados. Quando usamos a função `pd.read_csv(url)`, estamos trazendo uma tabela de dados da internet (ou do seu computador) para dentro do Python na forma de um **DataFrame** (que é o nome técnico para tabelas no Pandas).

````python
# Importa a biblioteca Pandas, que é usada para carregar, manipular e analisar tabelas de dados
import pandas as pd
````

````python
# Guarda o link da internet onde o arquivo CSV está salvo
url='https://raw.githubusercontent.com/alura-cursos/Pandas/refs/heads/main/superstore_data.csv'

# Lê o arquivo de texto (CSV) da internet e o transforma em uma tabela (DataFrame) do Pandas, indicando que as colunas são separadas por vírgula
dados_mercado=pd.read_csv(url,sep= ',')

# Mostra na tela apenas as 5 primeiras linhas da tabela para visualizarmos a estrutura
dados_mercado.head()
````

````python
# Lê o arquivo CSV da internet, mas carrega apenas as primeiras 10 linhas para economizar memória
dados_primeiras_linhas=pd.read_csv(url,sep= ',', nrows=10)

# Exibe a tabela gerada com as 10 linhas carregadas
dados_primeiras_linhas
````

````python
# Lê o arquivo CSV trazendo apenas as colunas específicas que passamos na lista: 'Id', 'Year_Birth' e 'Income'
dados_selecao=pd.read_csv(url,sep= ',', nrows=10,usecols=['Id','Year_Birth','Income'])

# Exibe a tabela filtrada
dados_selecao
````

````python
# Carrega o CSV trazendo apenas a primeira coluna (índice 0) e a quarta coluna (índice 3)
dados_selecao=pd.read_csv(url,sep= ',', nrows=10,usecols=[0,3])

# Exibe o resultado obtido
dados_selecao
````

### Parâmetros Avançados para Leitura de CSV

Ao ler arquivos CSV, é comum precisar de mais controle sobre o processo. Alguns parâmetros importantes incluem:

*   `skiprows`: Permite pular um número específico de linhas do início do arquivo (por exemplo, cabeçalhos ou metadados).
*   `skipfooter`: Permite pular um número específico de linhas do final do arquivo (útil para rodapés ou comentários). **Nota**: `skipfooter` geralmente requer o `engine='python'` para funcionar corretamente, pois o motor C padrão do Pandas não suporta essa funcionalidade.
*   `encoding`: Define a codificação de caracteres do arquivo (ex: 'ISO-8859-1', 'utf-8'). Arquivos com codificações diferentes podem causar erros na leitura se o `encoding` não for especificado corretamente.
*   `sep`: Define o delimitador de colunas no arquivo CSV (por exemplo, ',' para vírgula, ';' para ponto e vírgula).

### ⚙️ Parâmetros importantes de leitura

Quando os dados são muito grandes ou possuem formatos especiais, usamos alguns ajustes de configuração:
* **`skiprows`**: Pula linhas inúteis no início do arquivo (como cabeçalhos de relatórios).
* **`skipfooter`**: Ignora observações ou resumos ao final do arquivo.
* **`encoding`**: Ajuda o Python a ler acentos e caracteres especiais da língua portuguesa corretamente (ex: `ISO-8859-1`).

````python
# URL contendo os dados do SUS no GitHub
url='https://raw.githubusercontent.com/alura-cursos/Pandas/main/dados_sus.csv'

# Carrega os dados com configurações avançadas:
# - encoding='ISO-8859-1' para ler caracteres especiais em português corretamente
# - skiprows=3 para pular as 3 primeiras linhas de cabeçalho do arquivo
# - skipfooter=9 para ignorar as últimas 9 linhas de rodapé do arquivo
# - sep=';' define o ponto e vírgula como separador
# - engine='python' é necessário para poder usar o skipfooter
dados=pd.read_csv(url,encoding='ISO-8859-1',skiprows=3,skipfooter=9,sep=';',engine='python')

# Exibe as 5 primeiras linhas da tabela processada
dados.head()
````

### Lendo Arquivos Excel

Esta seção aborda a leitura de dados de arquivos Excel (`.xlsx` ou `.xls`) para DataFrames Pandas. Arquivos Excel são frequentemente usados para armazenamento de dados tabulares e relatórios.

### 📊 Lendo tabelas do Excel (`.xlsx`)

Arquivos Excel diferem de arquivos CSV por poderem ter múltiplas abas (planilhas internas). No Pandas, usamos `pd.read_excel()` para carregar estes arquivos. Caso queira escolher uma aba específica, basta usar o parâmetro `sheet_name` informando o nome da aba desejada.

### Dica para Acesso a Arquivos "Raw" no GitHub

Ao tentar ler arquivos diretamente do GitHub (ou outros repositórios Git), a URL padrão pode não ser a URL do arquivo "raw". Para garantir que o Pandas (ou outras bibliotecas) consiga ler o conteúdo do arquivo corretamente, adicione `?raw=True` ao final da URL do arquivo no GitHub. Isso força a entrega do conteúdo bruto do arquivo.

````python
# Armazena o link do arquivo Excel hospedado no GitHub, usando o final '?raw=True' para baixar o arquivo correto
url='https://github.com/alura-cursos/Pandas/blob/main/emissoes_CO2.xlsx?raw=True'
````

````python
# Usa o Pandas para ler o arquivo Excel carregado a partir do link fornecido
dados_co2=pd.read_excel(url)
````

````python
# Exibe as primeiras 5 linhas da tabela de emissões de CO2 carregada do Excel
dados_co2.head()
````

### Verificando Múltiplas Abas em Arquivos Excel

É possível carregar apenas um subconjunto de colunas de uma aba específica de um arquivo Excel usando o parâmetro `usecols`. Você pode especificar as colunas por nome ou por um intervalo de colunas no formato Excel (ex: 'A:D').

````python
# Lê a aba 'emissoes_C02' trazendo apenas o intervalo de colunas de 'A' até 'D'
intervalo=pd.read_excel(url,sheet_name='emissoes_C02',usecols='A:D')

# Mostra as primeiras 5 linhas com apenas esse intervalo selecionado
intervalo.head()
````

````python
# Lê apenas o intervalo de colunas de 'A' até 'D' e limita o carregamento para apenas as primeiras 10 linhas da tabela
intervalo_2=pd.read_excel(url,sheet_name='emissoes_C02',usecols='A:D',nrows=10)

# Mostra o resultado final limitado
intervalo_2.head()
````

### Trabalhando com Dados JSON

JSON (JavaScript Object Notation) é um formato de dados leve e muito utilizado para troca de informações entre sistemas. Esta seção aborda como ler e manipular dados no formato JSON com Pandas.

### 🕸️ O que é JSON e como lê-lo?

O **JSON** é um formato leve e muito comum para transferir dados na internet. Muitas vezes os dados vêm aninhados (uma informação dentro da outra, como um dicionário). Para converter de forma plana em uma tabela limpa com linhas e colunas estruturadas, usamos o recurso **`pd.json_normalize()`**.

````python
# Importação de segurança da biblioteca Pandas para garantir que esteja ativa
import pandas as pd
````

````python
# Lê o arquivo de dados em formato JSON e o converte para uma tabela utilizável
dados_paciente=pd.read_json("/content/pacientes.json")

# Exibe toda a tabela de pacientes na tela
dados_paciente
````

````python
# Carrega um segundo arquivo JSON que possui uma estrutura interna aninhada (objetos dentro de objetos)
dados_paciente_2=pd.read_json("/content/pacientes_2.json")

# Exibe a tabela mostrando como as informações internas ficam compactadas em uma única coluna
dados_paciente_2
````

### Normalizando Dados JSON Aninhados

Frequentemente, dados JSON vêm com estruturas aninhadas (objetos dentro de outros objetos). Para transformar esses dados aninhados em uma estrutura tabular plana adequada para DataFrames Pandas, usamos a função `pd.json_normalize()`. Isso é essencial para facilitar a análise de dados complexos.

````python
# Desmembra a coluna 'Pacientes' que continha dados aninhados (dicionários) e os espalha em colunas individuais organizadas
df_dados=pd.json_normalize(dados_paciente_2["Pacientes"])

# Exibe a tabela normalizada de forma limpa e organizada
df_dados
````

### Salvando DataFrames em Arquivos JSON

Após processar e normalizar dados JSON, pode ser necessário salvá-los de volta em um arquivo JSON. O método `.to_json()` do Pandas permite exportar um DataFrame para esse formato.

````python
# Salva o DataFrame normalizado em um arquivo no formato JSON no ambiente local
df_dados.to_json("dados_normalizados.json")
````

### Consumindo APIs com Respostas JSON

Muitos serviços web (APIs) retornam dados no formato JSON. Esta seção mostra como fazer requisições a uma API, obter a resposta em JSON e convertê-la em um formato utilizável pelo Pandas.

````python
# Importa a biblioteca 'requests' para fazer requisições a sites e APIs na web
import requests

# Importa a biblioteca padrão 'json' para lidar com estruturas de dados JSON do Python
import json
````

### Obtendo Dados de uma API Externa com `requests.get()`

Para interagir com APIs RESTful e obter dados, usamos a biblioteca `requests` do Python, especificamente a função `requests.get()`. Basta passar a URL do endpoint da API como argumento para essa função.

````python
# Faz uma requisição GET para acessar os dados de usuários de uma API de testes pública
dados_usuarios = requests.get('https://jsonplaceholder.typicode.com/users')
````

### Convertendo Resposta de API para Objeto Python (JSON)

A resposta de uma requisição `requests.get()` geralmente vem como uma string de texto. Se a API retorna JSON, precisamos converter essa string JSON em um objeto Python (como um dicionário ou lista de dicionários) usando `json.loads()`. O atributo `.text` do objeto de resposta `dados_usuarios` contém a string JSON.

````python
# Converte o texto retornado da API (que é um JSON bruto) para um objeto de listas e dicionários nativos do Python
resultado = json.loads(dados_usuarios.text)
````

### Visualizando os Dados da API em um DataFrame

Após converter a resposta JSON da API para um objeto Python, podemos facilmente transformá-lo em um DataFrame Pandas para visualização e análise.

````python
# Transforma o objeto Python obtido da API diretamente em uma tabela do Pandas
pd.DataFrame(resultado)
````

### Prática: Normalizando Dados JSON de API

Esta seção é um exercício para praticar a normalização de dados JSON obtidos de uma API. A normalização é fundamental para transformar estruturas de dados complexas em tabelas planas para análise.

````python
# Faz uma nova requisição à API para baixar as informações atualizadas de usuários
dados=requests.get("https://jsonplaceholder.typicode.com/users")
````

````python
# Converte o texto cru da resposta JSON obtido da API em uma lista/dicionário Python estruturado
resultado = json.loads(dados_usuarios.text)
````

### Dica de Normalização: Usando `sep='_'` em `pd.json_normalize()`

Ao normalizar JSONs com `pd.json_normalize()`, o argumento `sep='_'` é extremamente útil. Ele define o separador para os nomes das colunas aninhadas. Por exemplo, se você tem `{"address": {"street": "Main"}}`, com `sep='_'` a coluna se tornaria `address_street`. Isso ajuda a manter a clareza e a evitar colisões de nomes, tornando os dados mais fáceis de acessar e manipular.

````python
# Desmembra o JSON aninhado criando colunas separadas por Underline (_) para os campos internos (como rua, cidade e geolocalização)
normal=pd.json_normalize(resultado,sep="_")

# Exibe o DataFrame resultante plano e estruturado
normal
````

## Trabalhando com Bancos de Dados

Esta seção introduz a interação com bancos de dados, focando em como criar, popular, consultar e manipular dados em um banco de dados SQL usando Pandas e SQLAlchemy.

### 🗄️ O que é um Banco de Dados Relacional e SQL?

Para trabalharmos com grandes volumes de dados de forma segura, guardamos as informações em bancos de dados. Com o auxílio do **SQLAlchemy**, conseguimos criar conexões e usar a linguagem **SQL** para interagir diretamente com as tabelas, permitindo fazer:
* **`SELECT`**: Consultar e trazer dados específicos.
* **`INSERT`**: Adicionar novos registros.
* **`UPDATE`**: Atualizar informações existentes.
* **`DELETE`**: Apagar registros.

### Criando um Banco de Dados SQLite em Memória

Usaremos a biblioteca SQLAlchemy para criar um banco de dados SQLite temporário, que reside apenas na memória (`:memory:`). Isso é ideal para testes e exemplos rápidos, pois o banco de dados é destruído quando a sessão é encerrada.

````python
### SQL: Structured Query Language

SQL (Structured Query Language) é a linguagem padrão para gerenciar e manipular bancos de dados relacionais. É usada para consultar, atualizar, inserir e deletar dados, além de gerenciar a estrutura do banco de dados.
````

````python
# Importa a biblioteca SQLAlchemy para conexão com bancos de dados relacionais
import sqlalchemy

# Importa a classe 'text' usada para escrever queries SQL puras de maneira segura
from sqlalchemy import text
````

````python
# Importa as classes do SQLAlchemy para criar a conexão e inspecionar o banco
from sqlalchemy import (create_engine,# motor oq faz o banco de dados funcionar
 MetaData,# classe de metadados que representa o banco de dados
  Table,# classe que represneta tabela em um banco de dados de uma tabela especifica
   inspect)# inspecionar o banco de dados
````

````python
# Cria um motor (Engine) para gerenciar um banco de dados SQLite temporário e rápido que roda na memória RAM do computador
engine= create_engine("sqlite:///:memory")
````

### Escrevendo um DataFrame em um Banco de Dados Local

O Pandas, em conjunto com SQLAlchemy, facilita a escrita de um DataFrame inteiro para uma tabela em um banco de dados. O método `.to_sql()` é usado para essa finalidade.

````python
# Guarda a URL direta para o arquivo CSV de clientes de banco hospedado no GitHub
url='https://raw.githubusercontent.com/alura-cursos/Pandas/refs/heads/main/clientes_banco.csv'
````

### Verificando Alterações no Banco de Dados após Operações DML

Após realizar operações de DML (Data Manipulation Language) como `UPDATE` ou `DELETE` diretamente no banco de dados via SQLAlchemy, os DataFrames que foram carregados *antes* dessas operações não refletem automaticamente as mudanças. É necessário executar uma nova consulta para carregar os dados atualizados do banco de dados e verificar as alterações.

````python
# Consulta SQL de teste para trazer informações de ID, escolaridade e rendimento dos IDs alterados ou apagados
query_check = "SELECT ID_Cliente, Grau_escolaridade, Rendimento_anual FROM clientes WHERE ID_Cliente IN (5008808, 6840104, 5008804)"

# Carrega os resultados do banco em formato de DataFrame do Pandas
updated_clientes = pd.read_sql(query_check, engine)

# Exibe o resultado das alterações
display(updated_clientes)

# Nota: Você deve ver 'Ensino superior' para ID_Cliente 5008808 e '300000.0' para Rendimento_anual de ID_Cliente 6840104
# ID_Cliente 5008804 não deve aparecer caso o DELETE tenha sido efetuado com sucesso
````

````python
# Define o comando em SQL para excluir o cliente com ID igual a 5008809 da tabela 'clienta'
deletar= 'DELETE FROM clienta WHERE ID_Cliente=5008809'

# Conecta ao banco de dados, executa a exclusão de registro e realiza commit das alterações
with engine.connect() as conn:
  result=conn.execute(text(deletar))
  conn.commit()
````

### O que é um Banco de Dados?

Bancos de dados são sistemas que permitem o armazenamento, organização e recuperação de informações de forma estruturada e eficiente. Eles são amplamente utilizados em muitas áreas, desde gerenciamento de negócios à pesquisa científica. Um banco de dados bem projetado pode melhorar significativamente a eficiência e a precisão das operações de uma organização.

### Principais Cláusulas SQL (DQL - Data Query Language)

As instruções SQL para consulta de dados (DQL) são compostas por cláusulas que fornecem informações detalhadas sobre como os dados devem ser recuperados. As cláusulas mais comuns são:

*   `SELECT`: **O que** queremos ver. Especifica quais colunas devem ser selecionadas na consulta.
*   `FROM`: **De onde** vêm os dados. Especifica as tabelas do banco de dados que devem ser consultadas.
*   `WHERE`: **Quais** condições os dados devem atender. Filtra os resultados da consulta com base em uma ou mais condições especificadas.
*   `ORDER BY`: **Como** os resultados devem ser ordenados. Classifica os resultados da consulta em ordem crescente (ASC) ou decrescente (DESC) com base em uma ou mais colunas.
*   `GROUP BY`: **Como** os dados devem ser agrupados. Agrupa os resultados da consulta com base em uma ou mais colunas, frequentemente usado com funções de agregação (COUNT, SUM, AVG, etc.).
*   `LIMIT`: **Quantos** resultados queremos. Limita o número de linhas retornadas pelos resultados da consulta.
