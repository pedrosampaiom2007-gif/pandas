### 1. Os principais padrões de regex

| Padrão | O que significa | Exemplo |
| :--- | :--- | :--- |
| `\d` | Um número (0–9) | 5 |
| `\d+` | Um ou mais números | 210 |
| `\s` | Um espaço em branco | |
| `\s+` | Um ou mais espaços seguidos | |
| `\w` | Letra, número ou _ | Pedro_2 |
| `.` | Qualquer caractere, geralmente exceto quebra de linha | a, 7, @ |
| `+` | Um ou mais do padrão anterior | `a+` encontra `aaa` |
| `*` | Zero ou mais do padrão anterior | `a*` encontra `aaa` ou nada |
| `^` | Início do texto | `^Pedro` |
| `$` | Final do texto | `210$` |

### 2. Mais 4 que valem a pena conhecer

| Padrão | Significado |
| :--- | :--- |
| `[abc]` | Encontra a, b ou c |
| `[0-9]` | Um dígito de 0 a 9 |
| `[A-Z]` | Uma letra maiúscula de A a Z |
| `\.` | Um ponto literal `.` |


# Exercícios da aula


```python
dados= pd.read_json('/content/dados_hospedagem.json')
```

```python
dados.head()
```

```python
# Normalizando os dados, arrumando aquele treco feio chei de dicionario, virou um data frame

dados= pd.json_normalize(dados['info_moveis'])
dados
```

### Removendo as listas dentro  das colunas

```python
# metodo explode faz listas virarem linhas no data frame
```

```python
colunas= list(dados.columns)
colunas
```

```python
dados=dados.explode(colunas[3:])# armazenamos os dados em uma lista e colocamos pra explodir
dados
```

```python
# deu problema nos indices depois que usamos o explode(lista virou linha) então vamos resetar os indices  pra ficar 0,1,2,3 etc...

dados.reset_index(inplace=True,drop=True)# inplace true garante que vamos mudar direto no data frame, o drop =true faz com que os indices  antigos errados saiam
```

```python
dados.info()# verificando tipo de dado,

# muitos dados object que se identificam assim, quando: há varios tipos de dados diferentes, um tipo especifico que nao foi identificado, mas todos são strings
```

```python
# Convertendo dados antes objects para numéricos usando numpy, vc escolhe o tipo de dado e poe no parentesis
# atribuindo um novo tipo de dado a coluna, essa é a sintaxe
dados['max_hospedes']=dados['max_hospedes'].astype(np.int64)

```

```python
col_numericas=['quantidade_banheiros','quantidade_quartos','quantidade_camas']
```

```python
dados[col_numericas]=dados[col_numericas].astype(np.int64)# transformamos os dados todos pra int
```

```python
dados['avaliacao_geral']=dados['avaliacao_geral'].astype(np.float64)# transformando os dados da coluna  de object pra float 64
```

```python
dados.info()
```

```python
# transformando a coluna dos $ que atrapalham o tratamento dos dados  através do APPLY LAMBDA

dados['preco']=dados['preco'].apply(lambda x: x.replace('$'," ").replace(","," ").strip())

# pra remover o $ e substitui-lo usamos o .apply(lambda) e colocamos o replace, a primeira string é o que vai ser substituído
# a segunda é o que vai entrar no lugar do que foi substituido no caso vamos por nada " ".strip é pra tirar os espaços em branco
```

```python
# Corrigindo a limpeza removendo os espaços em branco que surgiram da substituição da vírgula
dados['preco'] = dados['preco'].apply(lambda x: x.replace(' ', ''))
dados['preco'] = dados['preco'].astype(np.float64)
```

```python
dados['preco']
```

```python
#transformando mais de uma coluna de uma vez

# Substituindo '$' e ',' por nada, e removendo espaços, salvando o resultado de volta no DataFrame
dados[['taxa_deposito','taxa_limpeza']] = dados[['taxa_deposito','taxa_limpeza']].map(lambda x: x.replace(' ', '').replace('$', '').replace(',', ''))

# Convertendo para float64
dados[['taxa_deposito','taxa_limpeza']] = dados[['taxa_deposito','taxa_limpeza']].astype(np.float64)
```

# tokenização  ocorre quando dividimos o texto em partes menores ou tratamos o texto, exemplo: eu amo minha namorada= ['eu','amo','minha','namorada']

```python
# 1. Recuperamos a coluna original do arquivo original para limpar qualquer NaN anterior
dados_original = pd.read_json('/content/dados_hospedagem.json')
dados_original = pd.json_normalize(dados_original['info_moveis'])
dados_original = dados_original.explode(colunas[3:])
dados_original.reset_index(inplace=True, drop=True)

# Restauramos os textos originais tratando nulos
dados['descricao_local'] = dados_original['descricao_local'].fillna("")
dados['descricao_local'] = dados['descricao_local'].str.lower() # deixamos tudo em letra minúscula
```

```python
# removendo caracteres com regex
# Usamos r'...' para definir uma string bruta (raw string) e tratar os caracteres especiais com segurança
dados['descricao_local'] = dados['descricao_local'].str.replace(r'[^a-zA-Z0-9]', ' ', regex=True)
```

```python
# remover hifen solto
# Usamos r'...' para evitar avisos de caracteres de escape inválidos no regex
dados['descricao_local'] = dados['descricao_local'].str.replace(r'(?<!\w)-(?!\w)', ' ', regex=True)
```

```python
# Criando token usando split(divide elementos e transforma em lista)
dados['descricao_local'].str.split()
```

```python
#removendo {} e as " substituindo por nada

dados['comodidades']=dados['comodidades'].str.replace('\{|}|\"','',regex=True)
```

```python
dados['comodidades']=dados['comodidades'].str.split(',')
dados.head()
```
